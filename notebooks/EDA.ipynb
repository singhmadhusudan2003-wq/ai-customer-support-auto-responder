# Exploratory Data Analysis
## AI Customer Support Auto-Responder — Dataset EDA

Analyzes `dataset/customer_support_dataset.csv` (10,500 synthetic customer queries) across intent and sentiment distributions, query length, and class balance ahead of model training.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../dataset/customer_support_dataset.csv")
print(f"Shape: {df.shape}")
df.head()
```

```python
df.isnull().sum()
```

## Intent distribution

```python
intent_counts = df["intent"].value_counts()
print(intent_counts)
intent_counts.plot(kind="bar", title="Query count by intent", color="#4f46e5")
plt.ylabel("count")
plt.tight_layout()
plt.show()
```

## Sentiment distribution

```python
sentiment_counts = df["sentiment"].value_counts()
print(sentiment_counts)
sentiment_counts.plot(kind="bar", title="Query count by sentiment", color="#f59e0b")
plt.ylabel("count")
plt.tight_layout()
plt.show()
```

## Intent x Sentiment cross-tabulation

```python
cross = pd.crosstab(df["intent"], df["sentiment"])
cross
```

```python
cross.plot(kind="bar", stacked=True, figsize=(9, 5), colormap="viridis")
plt.title("Intent vs Sentiment")
plt.ylabel("count")
plt.tight_layout()
plt.show()
```

## Query length distribution

```python
df["query_length"] = df["query"].str.split().apply(len)
df["query_length"].describe()
```

```python
df["query_length"].plot(kind="hist", bins=20, title="Query length (words)", color="#10b981")
plt.xlabel("words per query")
plt.tight_layout()
plt.show()
```

## Most common words per intent (simple term frequency)

```python
from collections import Counter
import re

stopwords = {"the", "a", "an", "my", "i", "is", "to", "for", "and", "of", "in", "you", "your", "it", "this"}

for intent in df["intent"].unique():
    text = " ".join(df[df["intent"] == intent]["query"]).lower()
    words = [w for w in re.findall(r"[a-z']+", text) if w not in stopwords]
    top = Counter(words).most_common(8)
    print(f"{intent}: {top}")
```

## Summary

- Dataset is well-balanced across the 6 intent classes and 3 sentiment classes (template-generation ensures no severe class imbalance).
- Query length is short (typically 8-20 words), consistent with real customer support chat messages.
- Distinct vocabulary per intent (e.g. "refund"/"money" for Refund, "password"/"account" for Account Issue) explains why even a simple TF-IDF + Logistic Regression model achieves very high accuracy — see `models/saved/metrics_logistic_regression.json`.
